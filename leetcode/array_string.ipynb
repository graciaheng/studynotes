## LeetCode 88 — Merge Sorted Array

**Pattern:** STL

### Key Insight
- Fill `nums1` from the back to avoid overwriting elements.
- since nums1 has a fixed size, there are already empty slots in there, hence if you append/extend in python, you would just be creating more array slots

### Complexity
- Time: O(m + n)
- Space: O(1)

### Java Solution
```java
class Solution {
    public void merge(int[] nums1, int m, int[] nums2, int n) {
        for (int j = 0, i = m; j < n; j++) {
            nums1[i] = nums2[j];
            i++;
        }
        Arrays.sort(nums1);
    }
}
```


````python
  
  def merge(self, nums1: list[int], m: int, nums2: list[int], n: int) -> None:
        """
        Do not return anything, modify nums1 in-place instead.
        """
        for j in range(n):
          nums1[m+j] = nums2[j] #copy into empty spaces
        nums1.sort()
````

## Leetcode 27 -- Remove Element

### Key Insights
- pointers in java are just indices

*when will i use pointers?*

- When you need to work from both ends
- When you need to filter or rearrange an array in place
- When searching a sorted array for a pair
- When you need to track different speeds or positions

### Complexity

- Time complexity: O(n)
- Space complexity: O(1)

### Java Solution
```java
class Solution {
    public int removeElement(int[] nums, int val) {
        // remove all that is val
        // k is numbers of elements that are not val

        int k = 0;

        for (int i = 0; i < nums.length; i++) {
            if (nums[i] != val) {
               nums[k] = nums[i]; //'pointer'
               k++;
            }
        }
        return k;
    }
}
```

### Leetcode 26 -- Remove Duplicates from Sorted Array I

### Key Insights
- pointers in java are just indices

*when will i use pointers?*

- as this is a sorted array, you only need to compare with the element before it
- consider if array is empty

### Complexity

- Time complexity: O(n)
- Space complexity: O(1)

### Java Solution
```java
class Solution {
    public int removeDuplicates(int[] nums) {
        //each element should only appear once
        //k is the number of unique elements
        //given array is sorted

        //if array is empty
        if (nums.length == 0) {
            return 0;
        }

        int k = 1;

        for (int i = 1; i < nums.length; i++) {
            //if this is a unique number
            if (nums[i] != nums[i - 1]) {
                nums[k] = nums[i];
                k++;
            }
        }

        return k;
    }
}
```

## Leetcode 26 -- Remove Duplicates from Sorted Array II

### Key Insights
- pointers in java are just indices
- i = where you are reading
- k = where you are writing / how many valid elements you've kept.
- k - 2 as we are checking against the current valid duplicates we have, not the ones from the original array

### Complexity

- Time complexity: O(n)
- Space complexity: O(1)

### Java Solution
```java
class Solution {
    public int removeDuplicates(int[] nums) {
        //each element only appears twice
        //k is number of unique elements

        //return this as we allow max 2 duplicates
        if (nums.length <= 2) {
            return nums.length;
        }

        int k = 2;

        for (int i = 2; i < nums.length; i++) {
            //does not equal the previous 2 elements before
            if (nums[i] != nums[k - 2]) {
                nums[k] = nums[i];
                k++;
            }
        }
        return k;
    }
}
```

## Leetcode 121 -- Best Time to buy and sell stock

### Key Insights
- why not a hashmap? the cheapest previous price is all you need to calculate the maximum profit
- dont need to store everything

### Complexity

- Time complexity: O(n)
- Space complexity: O(1)

### Java Solution
```java
class Solution {
    public int maxProfit(int[] prices) {
        //maximise profit 
        //choose the day with cheapest price & maximum price (after the day of cheapest price)

        int max = 0;
        int minPrice = Integer.MAX_VALUE;

        //getting max difference
        for (int i = 0; i < prices.length; i++) {
           if (max < prices[i] - minPrice) {
                max = prices[i] - minPrice;
           }          
           if (prices[i] < minPrice) {
                minPrice = prices[i];
           }
        }
        return max;
    }
}
```

## Leetcode 122 -- Best Time to buy and sell stock II

### Key Insights
- need to account for final day

### Complexity

- Time complexity: O(n)
- Space complexity: O(1)

### Java Solution
```java
class Solution {
    public int maxProfit(int[] prices) {
        int max = 0;
        int minPrice = prices[0];

        for (int i = 1; i < prices.length; i++) {
            if (prices[i] < prices[i - 1]) {
                max += prices[i - 1] - minPrice;
                minPrice = prices[i];
            }
        }
        //sell on final day
        max += prices[prices.length - 1] - minPrice;
        return max;
    }
}
```

## Leetcode 1 -- Two Sum

### Key Insights
- use HashMaps whe you need 2 values
- new int[] {} for creating a new array to be returned
- for this problem, you must put the indice value after you check, then it will be considered 
as proper checking, cause if you put it in before checking, means u include checking for that own value too

### Complexity

- Time complexity: O(n)
- Space complexity: O(n)

### Java Solution
```java
 public int[] twoSum(int[] nums, int target) {
        //2 of the nums must add up to target

        HashMap<Integer, Integer> indices = new HashMap<>();
    
        for (int i = 0; i < nums.length; i++) {
            if (indices.containsKey(target - nums[i])) {
                return new int[]{indices.get(target - nums[i]), i};
            }
            else {
                indices.put(nums[i], i);
            }
        }
        return new int[] {};
    }
```

## Leetcode 217 -- Contains Duplicate

### Key Insights
- HashSet is used as we only need to remember one value, since it is for ANY repeat

### Complexity

- Time complexity: O(n)
- Space complexity: O(n)

### Java Solution
```java
 class Solution {
    public boolean containsDuplicate(int[] nums) {
        //return any value that appears at least twice
        HashSet<Integer> duplicates = new HashSet<>();

        for (int i = 0; i < nums.length; i++) {
            //check for duplicates
            if (duplicates.contains(nums[i])) {
                return true;
            }
            duplicates.add(nums[i]);
        }
        return false;
    }
}
```

## Leetcode 219 -- Contains Duplicate II

### Key Insights
- HashMap is used as we need to remember the indice position

### Complexity

- Time complexity: O(n)
- Space complexity: O(n)

### Java Solution
```java
 class Solution {
    public boolean containsNearbyDuplicate(int[] nums, int k) {
        //checking if theres a duplicte within (i - j) <= k
        HashMap<Integer, Integer> duplicates = new HashMap<>();

        for (int i = 0; i < nums.length; i++) {
            if (duplicates.containsKey(nums[i])
            && (i - duplicates.get(nums[i])) <= k) {
                return true;
            }
            duplicates.put(nums[i], i);
        }

        return false;
    }
}

## Leetcode 242 -- Valid Anagram

### Key Insights
- to compare 2 hashmaps, use equals()
- getOrDefault(key, value)

### Complexity

- Time complexity: O(n)
- Space complexity: O(n)

### Java Solution
```java
class Solution {
    public boolean isAnagram(String s, String t) {

        if (s.length() != t.length()) {
            return false;
        }

        HashMap<Character, Integer> word1 = new HashMap<>();
        HashMap<Character, Integer> word2 = new HashMap<>();

        //to be an anagram, both must have the same amount of letters and the same type of letters

        for (char i : s.toCharArray()) {
            word1.put(i, word1.getOrDefault(i, 0) + 1);
        }

        for (char j : t.toCharArray()) {
            word2.put(j, word2.getOrDefault(j, 0) + 1);
        }

        if (word1.equals(word2)) {
            return true;
        }

        else {
            return false;
        }
    }
}
```

## Leetcode 49 -- Group Anagrams

### Key Insights
- use sorted and just compare if its an anagram from there
- anagrams.get(key).add(s); means to "Find the list stored under this key, then add the current word to that list."

### Complexity

- Time complexity: O(n)
- Space complexity: O(n)

### Java Solution
```java
class Solution {
    public List<List<String>> groupAnagrams(String[] strs) {
        //group anagrams in the same array in 2d array

        HashMap <String, List<String>> anagrams = new HashMap<>();

        for (String s : strs) {
            char[] chars = s.toCharArray();
            Arrays.sort(chars);

            String key = new String(chars);
            if (!anagrams.containsKey(key)) {
                anagrams.put(key, new ArrayList<>());
            }
            anagrams.get(key).add(s);
        }

        return new ArrayList<>(anagrams.values());   
    }
}
```

# Stack

## Leetcode 49 -- Group Anagrams

### Key Insights
- use a stack for matching pairs for most recent match
array: when you care about order/position
hashmap: when you need to remember information about something
hashset: you want to know whether something is already there 
stack: you need to deal with the most recent thing first (LIFO)

### Complexity

- Time complexity: O(n)
- Space complexity: O(n)

### Java Solution
```java
class Solution {
    public boolean isValid(String s) {
        Stack<Character> stack = new Stack<>();
        //push characters into stack
        for (char i : s.toCharArray()) {
            if (i == '(' || i == '{' || i == '[' ) {
                 stack.push(i);
            }

            else if (i == ')' || i == '}' || i == ']' ) {
                if (stack.isEmpty()) {
                    return false;
                }
                if (stack.peek() == '(' && i == ')'
                || stack.peek() == '{' && i == '}'
                || stack.peek() == '[' && i == ']') {
                    stack.pop();
                }
                else {
                    return false;
                }
            }
        }
        return stack.isEmpty();
    }
}

## Leetcode 71 -- Simplify Path

### Key Insights
- Break a large String into smaller pieces using split("/")
- Join a large string using String.join("/", stack);
dont forget edge cases like popping from an emptystack!!!!!!!!

### Complexity

- Time complexity: O(n)
- Space complexity: O(n)

### Java Solution
```java
class Solution {
    public String simplifyPath(String path) {
        Stack<String> stack = new Stack<String>();
        //stack.push("/"); //must start with single slash
        String[] parts = path.split("/");
        
        for (int i = 0; i < parts.length; i++) {
            if (parts[i].equals("..")) {
                if (!stack.isEmpty()) {
                    stack.pop();
                }
            }
            else if (parts[i].equals(".") || parts[i].equals("")) {
                continue;
            }
            else {
                stack.push(parts[i]);
            }
        }
        if (stack.isEmpty()) {
            return "/";
        }
        return "/" + String.join("/", stack);
    }
}
